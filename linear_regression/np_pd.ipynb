```python
def last_four(cc_num:str):
    cc_num = str(cc_num)
    return cc_num[-4:]
```

```python
df['CC Number'].apply(last_four)
```

```python
def quality(total_bill,tip):
    if tip / total_bill < 0.25:
        return tip / total_bill
    else:
        return "other"
```

```python
df[['total_bill','tip']].apply(lambda fd:quality(fd['total_bill'],fd['tip']),axis=1)
```
