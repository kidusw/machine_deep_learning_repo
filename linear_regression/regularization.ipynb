```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../3.1 UNZIP_ME_FOR_NOTEBOOKS_V4/DATA/Advertising.csv')
```

```python
X = df.drop('sales',axis=1)
y = df['sales']
```

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
poly_conv = PolynomialFeatures(degree=2)
```

```python
poly_features = poly_conv.fit_transform(X)

poly_features
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(poly_features, y, test_size=0.3, random_state=42)
```

```python
scaler.fit(X_train)
```

```python
X_train_scaled = scaler.transform(X_train)

X_test_scaled = scaler.transform(X_test)
```

```python
from sklearn.linear_model import RidgeCV
```

```python
ridge_cv = RidgeCV(alphas=(0.1,5,10))
```

```python
ridge_cv.fit(X_train_scaled,y_train)
```

```python
predictions = ridge_cv.predict(X_test_scaled)

predictions
```

```python
ridge_cv.coef_
```

```python
from sklearn.metrics import mean_squared_error
```

```python
mean_squared_error(y_test,predictions)
```

```python
from sklearn.linear_model import LassoCV
```

```python
lasso_cv = LassoCV(alphas=(0.1,5,10))
```

```python
lasso_cv.fit(X_train_scaled,y_train)
```

```python
lasso_predictions = lasso_cv.predict(X_test_scaled)

lasso_predictions
```

```python
lasso_cv.coef_
```

```python
mean_squared_error(y_test,lasso_predictions)
```

```python
from sklearn.linear_model import ElasticNetCV
```

```python
elastic_cv = ElasticNetCV(n_alphas=5,l1_ratio=[0.2,0.5,0.8])
```

```python
elastic_cv.fit(X_train_scaled,y_train)
```

```python
elastic_predictions = elastic_cv.predict(X_test_scaled)

elastic_predictions
```

```python
mean_squared_error(y_test,elastic_predictions)
```
