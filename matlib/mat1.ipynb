```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
x = np.arange(0,10)
y = x * 2
```

```python
plt.plot(x,y)
plt.xlabel('X values')
plt.ylabel('Y values')
plt.title("X Vs Y Graph")
plt.xlim(0,6)
```

```python
m = np.linspace(0,10,11)
c = 299_792_458 
E = m * c ** 2
E
```

```python
plt.plot(m,E,linewidth =8,color='red',linestyle='--')
plt.xlabel('Mass in Grams')
plt.ylabel('Energy in Joules')
plt.title('E=mc^2')
```

```python
E_log = np.log10(E)
plt.plot(m,E_log,linestyle='--',color='red');
```
